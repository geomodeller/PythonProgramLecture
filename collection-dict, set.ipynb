```python
my_lylics = """Yesterday, all my troubles seemed so far away
Now it looks as though they're here to stay
Oh, I believe in yesterday
Suddenly, I'm not half the man I used to be
There's a shadow hanging over me
Oh, yesterday came suddenly
Why she had to go
I don't know, she wouldn't say
I said something wrong
Now I long for yesterday
Yesterday, love was such an easy game to play
Now I need a place to hide away
Oh, I believe in yesterday
Why she had to go
I don't know, she wouldn't say
I said something wrong
Now I long for yesterday
Yesterday, love was such an easy game to play
Now I need a place to hide away
Oh, I believe in yesterday
"""
```

```python
alphabet_count = dict()
for character in my_lylics:
    # 영어인지 확인
    if character.isalpha():
        # 소문자 변환
        character = character.lower()
        # 값 세기 - 처음이면 1, 아니면 이전에 있던 수에 +1
        if character in alphabet_count:
            alphabet_count[character] += 1 # 이전에 있던 알파벳수 +1
        else:
            alphabet_count[character] = 1  # 처음 본 알파벳임으로 1로 설정
    else:
        # 영어가 아니라 무시하고 다음으로 진행
        continue
```

```python
alphabet_count_keys = alphabet_count.keys()
alphabet_count_keys = sorted(alphabet_count_keys)
for key in alphabet_count_keys:
    print(f'{key}는 {alphabet_count[key]:4}번 반복됨')
```
