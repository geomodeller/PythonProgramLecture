```python
def check_if_adult(age: int)->bool:
    """어른인지 확인하는 함수

    Args:
        age (int): 입장객의 나이

    Returns:
        bool: 어른이면 True, 아니면 False
    """
    if age > 18:
        return True
    else:
        return False
```

```python
visitor = [45, 35, 7, 10, 50, 29]
for i in filter(check_if_adult, visitor):
    print(i)
```

```python
def discount(p):
    return p*0.7

price_s = [300, 400, 600, 800, 1200]
for price in map(discount, price_s):
    print(price)
```

```python
def larger_than_three_letter(word):
    if len(word)>5:
        return True
    else:
        return False
my_list = ['apple', 'banana', 'mango', 'pineapple', 'kiwi', 'orange']
for fruit in filter(larger_than_three_letter, my_list):
    print(fruit)
```

```python
def hypotenuse(a, b):
    return (a**2 + b**2)**(1/2)

hypotenuse_2 = lambda a,b: (a**2 + b**2)**(1/2)


print(hypotenuse(3,4))
print(hypotenuse_2(3,4))
```
