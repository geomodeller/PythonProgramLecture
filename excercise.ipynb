```python
students = ["Alice", "Bob", "Charlie", "Diana", "Ethan", "Fiona"]
math_scores = [85, 92, 78, 90, 88, 75]
science_scores = [91, 89, 85, 94, 82, 79]
```

```python
# step 1: combine lists into a dictionary
student_data = dict()
for name, math, science in zip(students, math_scores, science_scores):
    student_data[name] = (math, science)
print(student_data)
```

```python
# step 2: find high performer (>90)
def check_high_score(name)->bool:
    scores = student_data[name]
    mean = sum(scores)/len(scores)
    if mean > 90:
        return True
    else:
        return False

high_performer = []
for name in filter(check_high_score, student_data.keys()):
    high_performer.append(name)

print(high_performer)
```

```python
# step 3: add english socre
english_scores = [88, 85, 82, 91, 87, 80]

for name, english in zip(student_data.keys(), english_scores):
    student_data[name] = student_data[name] + (english,)
    
print(student_data)
```

```python
# step 4: get the mean
average_score = dict() # or {}
for name, mean in map(lambda x: (x[0], sum(x[1])/len(x[1])), student_data.items()):
    average_score[name] = mean
```

```python
def calculate_mean(name)->float:
    scores = student_data[name]
    mean = sum(scores)/len(scores)
    return name, mean

average_score = dict() # or {}
for name, mean in map(calculate_mean, student_data.keys()):
    average_score[name] = mean
average_score
```

```python
# step 5: ranking
average_score = sorted(average_score.items(), key = lambda x:x[1], reverse = True)
for rank, (name, mean) in enumerate(average_score, 1):
    print(f'Rank {rank}: {name} - {mean}')
```
